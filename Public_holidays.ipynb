```python
import pandas as pd

df = pd.read_csv('/content/final_dataset_2022_2024.csv', index_col=0)
df.index = pd.to_datetime(df.index, dayfirst=True)
df = df.sort_index()

print(df.shape)
print(df.index.min(), "to", df.index.max())
```

```python
holiday_rows = df[df['is_public_holiday'] == 1].copy()
holidays_from_data = sorted(set(holiday_rows.index.date))

print(f"Found {len(holidays_from_data)} public holiday dates in your dataset:")
for d in holidays_from_data:
    print(d)
```

```python
future_holidays_2025_2026 = [
    '2025-01-01', '2025-02-03', '2025-03-17', '2025-04-21', '2025-05-05',
    '2025-06-02', '2025-08-04', '2025-10-27', '2025-12-25', '2025-12-26',
    '2026-01-01', '2026-02-02', '2026-03-17', '2026-04-06', '2026-05-04',
    '2026-06-01', '2026-08-03', '2026-10-26', '2026-12-25', '2026-12-28'
]

all_dates = sorted(set(holidays_from_data) | set(pd.to_datetime(future_holidays_2025_2026).date))
holidays_df = pd.DataFrame({'date': all_dates})
holidays_df.to_csv('/content/public_holidays_ie.csv', index=False)

print(f"Saved {len(holidays_df)} total holiday dates")
holidays_df.head(10)
```
